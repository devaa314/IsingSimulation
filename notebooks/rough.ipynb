# The Ising Model
## 1. Magnetic Phase Transitions
The magnetisation of certain materials, such as iron, follows 2 different forms of magnetisation. Paramagnetism at high temperatures ($T > T_c$ for some critical temperature $T_c$) and ferromagnetism at low temperatures ($T < T_c$). At high enough temperatures, the magnetisation $\vec{M}$ is roughly proportional to $\vec{B}$, and continuity holds at $\vec{B} = 0$, that is:

$$
\lim_{\vec{B}\rightarrow0}\vec{M} = 0
$$

So, as the magentic field goes to 0, so does magnetisation. However, if $T < T_c$, we actually have a discontinuity at $\vec{B} = 0$. Indeed, if we assume for simplicity $\vec{B} = B\hat{z}$, then we instead get:

$$
\lim_{B\downarrow{0}}M = M^{+}\\
\lim_{B\uparrow{0}}M = M^{-}
$$

where $M^+$ is some finite positive value and $M^-$ is some finite negative value. This is a clear discontinuity in $M$ for temperatures below $T_c$. Hence, we denote the transition from ferromagnetic to paramagnetic as $T$ reaches $T_c$ a *magnetic phase transition.*

There are some properties of this discontinuity. Firstly, the magnitude of the discontinuity, $M^+ - M^-$, decreases as $T \rightarrow T_c$ and eventually becomes 0, that is vanishes, when $T = T_c$. Furthermore, is $B = 0$ and $T$ is reduced to a value below $T_c$, we observe *spontaneous magnetisation*, that is, the material suddenly goes from 0 magnetisation to a finite magnetisation. The exact magnitude and sign is determined at random depending on the distribution of spins when $T$ drops below $T_c$. As we reduce $T$ further, the magnetisation finally saturates, attaining a certain largest positive or negative value.

This change from paramagnetic to ferromagentic behaviour can perhaps be explained as follows. The atoms in the material attempt to align with neighbours through exchange interactions. However, at higher temperatures, thermal fluctuations overwhelm this process, resulting in an essentially random allignment of spins so no permanent magnetisation is observed. However at lower temperatures, the exchange interactions win, so groups of alligned spins form forcing the system as a whole to have a total magentisation.

As we go from high temperatures to low temperatures, the temperature rises smoothly from zero magnetisation to non-zero magnetisation. This is a discontinuity in the derivative of $M$ with respect to $T$ so it is known as a *continuous phase transition* ot *second-order phase transition*.

Another continuous phase transition is seen in magnetic susceptibility $\chi$, given by:

$$
\chi = \frac{\partial{M}}{\partial{B}}
$$

The presence of the discontinuity in $M$ at $T_c$ indicates that $\chi$ must have infinite derivative at $T_c$.

## 2 The Ising Model of Magnetism
The magnetic moments of the particles in the material are of interest in this model. The protons and neutrons have negligible magnetic moment compared to the electrons since they are so much heavier. So they may be safely ignored. Hence, we model the material as an evenly spaced lattice of electrons. It point in the lattice is assgined it's electron spin, $s_i$. In the Ising model, the $s_i$ may adopt two values, namely, $\pm{1}$, spin up or down.

The next step is defining a Hamiltonian. They are really 2 types of terms in this Hamiltonian, the energy associated with each spin alligning or antialligning with the exeternal field and the energy associated with the interaction between the spins.

First we deal with the first type. Consider an arbritrary spin $s_i$ in an external field $B$. Then, the Hamiltonian for this situation is

$$
H = -\mu_BBs_i
$$

where $\mu_B$ is the Bohr magneton. This simply the energy of a magentic dipole in an external magentic field when the magnetic moment is parallel to the magentic field. Since lower energy states are always preferred, it is clear that the spin that is alligned with the external field is preferred.

Now we must account for energy associated between the interaction of the different spins. Suppose $s_i$ and $s_j$ are neighbouring spins. if the interaction energy between them is some constant $J$, the Hamiltonian between them is given by:

$$
H = -Js_is_j
$$

which is the interaction energy between 2 magnetic dipoles. The sign of $J$ is significant. If it is positive, alligned spins are prefered. if it is negative, antialligned spins are prefered. If $J > 0$, then the situation is known as *ferromagnetic* and if $J < 0$, it is *antiferromagnetic*. Therefore the total Hamiltonian for just the pair of spins is

$$
H = -Js_is_j -\mu_BBs_i
$$

For a general system with n spins, we have

$$
H = -J\sum_{(i, j)}s_is_j - \mu_bB\sum_{i = 1}^n s_i
$$

where $(i,j)$ merely ranges over nearest neighbour pairs since next-nearest neighbour interactions and beyond are generally much weaker and can be safely ignored. Certainly, we may also account for three-spin and four-spi interactions and beyond but in order to keep the model simple, these will be ignored.

## 3 Introduction of Statistical Mechanics
As with any system consisting of a large number of particles, it is fruitless to analyse macroscopic properties of the system (such as magnetisation) by working with individual microstates. This is where statitical mechanics comes in. If we assume the material sits at a temperature $T$, we may make use of the Boltzmann distribution to say the energy of any particular electron is given by the probability distribution:

$$
P(E_n) \propto e^{-E_n/k_BT}
$$

this of course prompts the introduction of the partition function $Z$:

$$
P(E_n) = \frac{e^{-E_n/k_BT}}{Z}
$$

The partition function is of utmost importance and will serve as the object of calculation moving forward as it containus all the thermodynamic information about the system. In fact, when it comes to the Ising model, solving the system often boils down to finding $Z$. This is because we may use $Z$ to find important thermodynamic quantities such as the free energy ($F$), ineternal energy ($U$) and specific heat ($C$). Indeed,

$$
F = U - TS
$$

where $S$, entropy, is given by

$$
S = -k_B\sum{P_n\ln(P_n)} = \frac{-k_B}{Z}\sum{e^{-E_n/k_BT}(-E_n/k_BT-\ln{Z})}
$$

where the sum ranges over all possible states. So when subsituting into the expression for $F$, we get a relatively simple expression for F in terms of Z:

$$
F = -k_BT\ln{Z}
$$

## 4: Reduced Quantities and Notation
Since the inverse of thermal energy occurs so often, we give it a compact form:

$$
\beta = \frac{1}{k_bT}
$$

We then define the dimensionless *reduced Hamiltonian* $-\beta{H}$:

$$
-\beta{H} = \frac{J}{k_bT}\sum_{(i, j)}s_is_j + \frac{\mu_BB}{k_bT}\sum_{i = 1}^n s_i.
$$

We then define two dimensionless constants $K = J/k_bT$ and $h = \mu_BB/k_BT$. So

$$
-\beta{H} = K\sum_{(i, j)}s_is_j + h\sum_{i = 1}^n s_i.
$$

We choose to work this form of the Hamiltonian for simplicity. We see that since $K$ and $h$ are inversely proportional to $T$, we see that at high temperatures, interaction energy is much smaller than thermal and vice-versa at small temperatures which is consistent with our intution for the phase transition.

Further, when working with free energy, it would also be useful to not only work with a dimensionless quantity but to also work with an intensive one. $F$ is extensive, meaning it scales with the number of spins, so, if we also divide $F$ by $N$, we yield the free energy
*per spin*

$$
f = -\frac{\beta{F}}{N} = \frac{1}{N}\ln{Z}
$$

It would also be useful to introduce notation for each microstate of the system. We write this as $\{s_i\}$. Therefore, $Z$ is

$$
Z = \sum_{\{s_i\}}e^{-\beta{H(\{s_i\})}}
$$

## 5. Solving the Ising Model in 1D: The Transfer Matrix
In this presentation, we will be solving the original problem solved by Ernst Ising in his PhD thesis, that is, the nearest-neighbour Ising model for a $1$ x $\infty$ lattice in the absence of an external magnetic field. We do this by the transfer-matrix method, finding the partition function for the general $1$ x $n$ case and taking the limit as $n \rightarrow \infty$.

First, let us consider the $1$ x $3$ lattice with 3 spins with *periodic boundary conditions*. That is, we make the assumption that $s_1$ and $s_3$ are nearest neighbours and include a 2-spin interaction term between them in the Hamiltonian. This is because these boundary conditions make the computations simpler as $n$ gets large in constrast to *free boundary conditions*, in which we assume the two spins at each end of the lattice are really far apart and therefore we neglect their interaction.

So what is this is mystical transfer matrix and how do we intepret it? Let us go back to each term in the parition function $Z$ which is of the form:

$$
e^{-\beta H(s_1, s_2, s_3)}.
$$

In this situation, the reduced Hamiltonian has form

$$
-\beta H = K(s_1s_2 + s_1s_3 + s_2s_3)
$$

so the term in the parition function is a product:

$$
e^{-\beta H(s_1, s_2, s_3)} = e^{K(s_1s_2)}e^{K(s_1s_3)}e^{K(s_2s_3)}.
$$


Since each $s_i$ has 2 values which it may adopt, each product of the form $s_is_j$ also only has 2 values it may adopt, $\pm 1$. So, we may conveniently define a $2$-by-$2$ matrix, $T$, which we may identify with this operation:

$$
\begin{bmatrix}
  e^K & e^{-K}\\
  e^{-K} & e^K
\end{bmatrix}
$$

This matrix is essentially indexed by the values $s_i$ and $s_j$ take. To be concrete, we may assign the following indeces: the top-left term is indexed by $(-1,-1)$, bottom-left by $(1, -1)$, top-right by $(-1, 1)$ and bottom-right by $(1, 1)$.

This is our *transfer matrix*. Why is this useful? We may now write our partition function as

$$
Z = \sum_{\{s_i\}}T_{s_1,s_2}T_{s_1,s_3}T_{s_2,s_3}.
$$

In order to iterate over each state $\{s_i\}$, we can first fix $s_1$ and $s_2$ to some arbritrary spins, and then sum over the states of $s_3$. Then we can sum over the states of $s_2$ and then over $s_1$. So instead we may write:

$$
Z = \sum_{\{s_1\}}\sum_{\{s_2\}}\sum_{\{s_3\}}T_{s_1,s_2}T_{s_1,s_3}T_{s_2,s_3}.
$$

This is a finite summation, so we may freely change the order:

$$
Z = \sum_{\{s_1\}}\sum_{\{s_3\}}\sum_{\{s_2\}}T_{s_1,s_2}T_{s_1,s_3}T_{s_2,s_3}.
$$

At this point, the summation over $\{s_2\}$ can be realised as matrix multiplication, indeed, the summation simply equals

$$
\sum_{\{s_2\}}T_{s_1,s_2}T_{s_2,s_3} = T^2_{s_1,s_3}.
$$

Subsituting this back in and making use of the fact that $T$ is symmetric gives :

$$
Z = \sum_{\{s_1\}}\sum_{\{s_3\}}T^2_{s_1,s_3}T_{s_3,s_1}.
$$

We then see the summation over $s_3$ is another matrix multiplication, this time between $T^2$ and $T$ so

$$
Z = \sum_{\{s_1\}}T^3_{s_1, s_1}
$$

which is the sum of the diagonal terms of the matrix $T^3$. This the trace of $T^3$ and it is known to equal the sum of the eigenvalues of $T^3$. Of course, the eigenvalues of $T^3$ are the eigenvalues of $T$ cubed. So solving $\det(T - \lambda I) = 0$:

$$
\det(T - \lambda I) = (e^K - \lambda)^2 - e^{-2K} = 0\\
e^K - \lambda = \pm e^{-K}\\
\lambda = 8\cosh(K), 2\sinh(K)
$$

giving that

$$
Z = 8(\cosh^3(K) + \sinh^3(K))
$$

Denote the first eigenvalue $\lambda_+$ and the other $\lambda_-$. Alright, that is fun, but how do we take this result into the limit $n \rightarrow \infty$? First, consider the partition function in the case of $n$ spins.

$$
Z_n = \sum_{\{s_1\}}\sum_{\{s_2\}} ... \sum_{\{s_n\}}T_{s_1,s_2}T_{s_2,s_3}T_{s_3, s_4}...T_{s_{n-1}, s_{n}}T_{s_n, s_1}
$$

We may evalue this summation as follows. First, isolate the sum over $\{s_2\}$ as follows and evaluate it by realising it as matrix multiplication as before

$$
\sum_{\{s_2\}}T_{s_1,s_2}T_{s_2,s_3} = T^2_{s_1,s_3}.
$$

Then subbing back in gives

$$
Z_n = \sum_{\{s_1\}} ... \sum_{\{s_n\}} T^2_{s_1,s_3}T_{s_3, s_4}...T_{s_{n-1}, s_{n}}T_{s_n, s_1}.
$$

Now, we have a matrix-multiplication-like sum over $\{s_3\}$ with the factors $T^2_{s_1,s_3}T_{s_3, s_4}$ so

$$
\sum_{\{s_3\}}T^2_{s_1,s_3}T_{s_3,s_4} = T^3_{s_1,s_4}
$$

and we can substitute this back in:

$$
Z_n = \sum_{\{s_1\}} ... \sum_{\{s_n\}} T^3_{s_1,s_4}T_{s_4, s_5}...T_{s_{n-1}, s_{n}}T_{s_n, s_1}.
$$

We can then interatively proceed finishing after the case i = n until we have:

$$
Z_n = \sum_{\{s_1\}}T^n_{s_1, s_1} = \lambda_+^n + \lambda_-^n
$$

which is pleasantly simple. So the reduced free energy per spin in the case of $n$ spins with periodic boundary conditions is:

$$
f_n = \frac{1}{n}\ln(Z_n) = \frac{1}{n}\ln(\lambda_+^n + \lambda_-^n) = \ln\lambda_+ + \frac{1}{n}\ln\left(1 + \left(\frac{\lambda_-}{\lambda_+}\right)^n\right)
$$

then taking the limit as $n \rightarrow \infty$

$$
f = \ln2 + \ln(\cosh(K))
$$

where $(\lambda_-/\lambda_+)^n \rightarrow 0$ since $\lambda_- < \lambda_+$ . This agrees with the result given by free boundary conditions which gives credence to why we chose to use periodic boundary conditions. Free boundary conditions would've meant we had to sum over all matrix entries of $T^n$ which would've been much more cumbersome to scale with $n$.

The prior computation was in the case of no external magnetic field. Thankfully, the situation for a finite external field is a straightforward modification. First write

$$
-\beta H = K \sum_{i = 1}^n s_i s_{i + 1} + h \sum_{i = 1}^n s_i
$$

where $s_{N + 1}$ = $s_{1}$. The partition function is then given by

$$
Z = \sum_{\{s_1\}} \sum_{\{s_2\}} ...
 \sum_{\{s_n\}} \prod_{i = 1}^n \exp \left(K s_i s_{i + 1} + h s_i \right).
$$

At this point, we *could* simply take

$$
T_{s_i, s_j} = e^{K s_i s_j + h s_i}
$$

and call it a day. However, we can do one better. Indeed, what comes next is some sleight-of-hand. Notice how from each term of the form $K s_i s_{i + 1} + h s_i$, we can take out a factor of $(h/2)s_i$ and pass it to the previous term. That is:

$$
Z = \sum_{\{s_1\}} \sum_{\{s_2\}} ...
 \sum_{\{s_n\}} \prod_{i = 1}^n \exp \left(K s_i s_{i + 1} + (h/2)(s_i + s_{i + 1})\right).
$$

This rewriting may pose a problem when $i = 1$ or $i = n$ because where do this terms pass to and receive from respectively? As it turns out, to and from on another respectively. This is solved by periodic boundary conditons. We may pass $(1/2)s_1$ to the last term as $s_{n + 1} = s_1$. Now we may define the transfer matrix by:

$$
T_{s_i, s_j} = e^{K s_i s_j + (h/2)(s_1 + s_j)}
$$

and just like before, the parition becomes:

$$
Z = \sum_{\{s_1\}} T^n_{s_1, s_1}.
$$

That's all fine and dandy, but it is worth taking a step back and thinking about why we made this subtle rewriting. After all, why couldn't we just define

$$
T_{s_i, s_j} = e^{K s_i s_j + h s_i}
$$

? This a perfectly valid matrix and, in fact, we can still write the product as a product of this matrix's elements and we get the exact same result using it's eigenvalues instead. The answer lies in the fact that this form yields in an *assymetric* matrix. Indeed:

$$
T_{-1, 1} = e^{(-K - h)}\\
T_{1, -1} = e^{(-K + h)}.
$$

While having a symmetric matrix is by no means *necessary*, it is always nice, for many reasons. Symmetric matrices always have a an eigenbasis consisting of real eigenvalues. Further, calculating eigenvalues of a symmetric matrix is just that bit easier. Let's do just that.

Now computing the eigenvalues for the symmetric transfer matrix, we get

$$
(e^{K - h} - \lambda)(e^{K + h} - \lambda) - e^{-2K} = 0\\
e^{2K} - e^{K-h} \lambda - e^{K + h} \lambda + \lambda^2 - e^{-2K} = 0\\
\lambda^2 - 2e^K\cosh(h)\lambda + 2\sinh(2K) = 0
$$

giving the 2 eigenvalues

$$
\lambda_+ = e^K\cosh(h) +
\sqrt{e^{-2K} + e^{2K}\sinh(h)^2}
$$

$$
\lambda_- = e^K\cosh(h) -
\sqrt{e^{-2K} + e^{2K}\sinh(h)^2}
$$

with $\lambda_+$ being larger. So as before, the reduced free energy per spin is

$$
f = \ln \lambda_+ = \ln \left(e^K\cosh(h) + \sqrt{e^{-2K} + e^{2K}\sinh(h)^2}\right).
$$

This is wonderful progress. Now, let us extract various quantities of interest, such as the reduced magnetisation, $m$ and magnetic susceptibility, $\chi$.

$m$, at any given spin, is simply the average value of the reduced spin. Of course, since all spins are the same in this scenario, we need not distinguish between them. Hence

$$
m = <s_1>
$$

So we need to find the probability distribution function of $s_i$. That involes taking the probability distribution for any state and summing over every other spin. For a moment, assume we have a finite $n$ spins again. With this in mind, $<s_1>$ is

$$
<s_1> = \frac{1}{Z} \sum_{\{s_1\}} s_1 \left(\sum_{\{s_2\}} ... \sum_{\{s_n\}} e^{-\beta H(\{s_i\})}\right).
$$

Consider the terms present in the sum on the right. They are all of the form

$$
\exp({\pm Ks_1 + hs_1}) \prod_{i = 2}^{n-1} \left(\exp(\pm K \pm h)\right) \exp({\pm Ks_1 \pm h})
$$

so if we differentiate by $h$, we pull down a factor $s_1$ which is excatly what we want. So indeed, if we simply switch the order of the derivative and summation, we receive

$$
m = <s_1> = \frac{1}{Z} \frac{\partial Z}{\partial h} = \frac{\partial}{\partial h}(\ln Z) = \frac{\partial f}{\partial h}
$$

where $f$ is simply the free energy per site. So we can compute the relevant derivative to find

$$
m = \frac{e^K \sinh h}{\sqrt{e^{-2K} + e^{2K}(\sinh h)^2}}
$$

which is a very nice, continous function with no pathological behaviour of any kind. We can clearly see for any $K$, $h = 0$ implies zero magnetisation so the material is paramagnetic at all temperatures ... which is bad news. What we wanted is that at certain low enough temperature, $h = 0$ can yield non-zero magnetisations which our expression for $m$ clearly does not display.

Have we gone wrong somewhere? I mean of course we have, after all our model does not reflect reality. Turns out, our mathematics and assumptions are not the problem, but the dimensionality we chose to work in.

If we used the exact same approach to solve the analagous 2D problem, such as Lars Onsager, we would indeed yield a magnetic phase transition! However, such a treatment is simply not tameable within the confines of today's talk. While obtaining an analytic solution might be out of our reach today, there are a myriad of ways in which we can demonstrate the existence of the magentic phase transition. I proceed with 2. One mathematical and one computational.

## 6. Mean Field Theory in 2D
The first way I will show this is using a useful but perhaps inaccurate approximation known as *mean field theory*.

The approximation relies on a rather simple principle: that any single spin in the lattice is representative of every other spin in the lattice. As such, we can replace all the spins with their mean spin.

Let us consider the 2D Ising model not in the presence of an external magnetic field. Then the Hamiltonian is

$$
H = -J \sum_{\langle ij \rangle} s_i s_j
$$

where the sum is taken over all pairs of nearest neighbours. Then, pick an arbitrary spin $s_0$. Consider the nearest neighbours of that spin and denote them $s_1, s_2, s_3, s_4$. Then replace all of these with their average value, if we remember, this is simply $m$, the reduced magnetisation. So

$$
s_1 = s_2 = s_3 = s_4 = m
$$

hence the Hamiltonian for just that spin is

$$
H_0 = -4Jms_0 \\
-\beta H_0 = \frac{4J}{k_b T}ms_0.
$$

Now here's the punchline. Since all the spins are the same in mean field theory, $\langle s_0 \rangle = m$ as well. Then calculating $\langle s_0 \rangle$ using the Boltzmann distribution:

$$
\langle s_0 \rangle = \frac{1}{Z_0} \sum_{\{s_0\}} s_0 e^{-\beta H_0} = \tanh\left[\frac{4J}{k_b T}m\right]
$$

yielding the following equation for $m$:

$$
m = \tanh\left[\frac{4J}{k_b T}m\right].
$$

This equation is not possible to solve analytically which is actually great news! By analysing this equation, we see that $m = 0$ is the only solution if $4J/k_b T \leq 1$. But if $4J/k_b T > 1$, $m = 0$ is solution along with 2 other non-zero solutions, one positive and one negative. This is astounding! Our model did reproduce the right results! The system transitions to ferromagnetic behaviour below the criticial temperature $T_c$:

$$T_c = 4\frac{J}{k_b}$$

Now I employ Newton's method to solve the equation for $m$ for certain temperatures $T$ in units of $T_c$ less than $1$. First I import necessary libraries and define some global variables and functions

```python
%matplotlib inline
```

```python
import numpy as np # for maths
import matplotlib.pyplot as plt # for plotting some pretty graphs
from matplotlib import colors # for plotting the state evolution
import matplotlib.animation as animation # for making animations
```

```python
# define a critical Cauchy difference which the program stops when the Cauchy difference falls under this value
epsilon_crit = 1e-6

# list of temperatures which are below the critical temperature and an initial value for m
temperatures = np.abs(np.random.normal(1, 0.5, 50))
m_0 = 1

# define the auxiliary function and its derivative as a function of magnetisation
def f(m, t):
  return m - np.tanh(m / t)

def f_prime(m, t):
  return 1 - np.square(1 / np.cosh(m / t)) / t
```

Then I iterate over each temperature and approximate the positive root using Newton's method.

```python
# define array to store m's as function of t
magnetisations = []

# iterate over each temperature
for t in temperatures:
  # buffer to store new value for m and the previous value
  m_buffer = [0, 1]

  # iterate as many times until desired precision has been reached
  while np.abs(m_buffer[0] - m_buffer[1]) > epsilon_crit:
    # compute new value of m
    m_buffer[0] = m_buffer[1]
    m_buffer[1] = m_buffer[1] - (f(m_buffer[1], t)/f_prime(m_buffer[1], t))

  magnetisations.append(m_buffer[1])

# plotting reduced magnetisation against reduced temperature
plt.scatter(temperatures * 4, magnetisations, label="MFA Results")

# plotting exact Onsager solution
temperaturesExact = np.linspace(0, 2.269185314213022, 72)
plt.plot(temperaturesExact, (1 - (np.sinh(2 / temperaturesExact)) ** (-4)) ** (1 / 8), color="red", label="Exact Onsager Solution")
plt.plot(np.linspace(2.26918531421, 8, 72), np.zeros(72), color="red")

plt.title("Reduced Magnetisation ($m$) against Temperature ($T$)")
plt.ylabel("$m$")
plt.xlabel("$T$ in units of $J / k_b$")
plt.legend()
plt.savefig("magnetisationvsTemperatureMFA.png")
plt.show()
```

So indeed, below the critical temperature, the system acheives a finite, non-zero magnetisation and saturates as the system gets colder and colder. Brilliant!

However, there is one drawback of this approximation. It does demonstrate the phase transition which is great, however, it doesn't do so at the right critical temperature. The exact solution due to Onsager demonstrated that the phase transition occurs at

$$T_c \approx 2.27 \frac{J}{k_b}$$

so the mean field value is quite a ways off. No fear, I shall harness some computational power to demonstrate this today.

## 7. The Metropolis-Hastings Algorithm
The crux of finding our magnetisation is, as we have already highlighted, is the partition function $Z$, after all, this would give us the full form of the probability distribution of the microstates

$$p(s) = \frac{e^{-\beta H(s)}}{Z}$$

and enadbles to calculate magnetisation as a function of temeperature. However, doing this in 2D analytically is extremely difficult. While possible, it would not fit within the confines of this talk. As such, I approach the problem computationally. We *could* consider a large lattice and compute $Z$ directly. However, the number of microstates for even a relatively small number of spins is enormous and is too computationally expensive. What we can do instead is to consider a sample of microstates.

The difficulty lies in generating a sample that is representative of the actual system at a given temperature $T$. If we knew $Z$, we could sample with probability given by $p(s)$. But of course, we don't! So we need another way of generating a sample.

One way of doing this is a *Metropolis-Hastings algorithm*. This algorithm works by first starting with some intial microstate and then we inductively sample more microstates by flipping a single arbritrary spin with a certain probability in the previous microstate. That probability is given by:

$$
P(s \rightarrow s') = \min \left\{1, e^{\beta(H(s_0) - H(s_1))}\right\}
$$

where $s$ is the current microstate and $s'$ is the microstate after flipping the arbitrary spin.

```python
N = 50 # number of spins on each axis

num_iterations = 1000 # number of times sweeping over lattice or equivalently, size of good sample of microstates generated
num_temperatures  = 72 # number of temperatures we want to consider
temperatures = np.linspace(1.27, 3.27, num_temperatures) # generate temperatures in units of J/k_b

magnetisations = np.zeros(num_temperatures) # an array to store magnetisations as a function of temperature

s_0 = 2 * np.random.randint(0, 2, (N, N)) - 1 # an abritrary microstate with which we begin our simulation

temperatures_interest = [0, 35, 71] # indeces of temperaturs which we want to track the evolution

state_evolution = np.empty((len(temperatures_interest), 1000, N, N)) # variable to store a state's evolution as a function of number of flips
```

At this point, we will want to iterate over each spin in the microstate and flip it with some probability. Suppose the state with the flipped spin is $s_1$. The the probability we do flip is given by:

$$
P(s_0 \rightarrow s_1) = \min \left\{1, e^{\beta(H(s_0) - H(s_1))}\right\}
$$

We can immediately see if flipping lowers the energy, then we always flip. However, if flipping increases the energy, then we only do so with probabilty $e^{\beta(H(s_0) - H(s_1))}$. This qualitatively makes sense. After all, on an individual microstate level, microstates of less energy are always more probable that microstates with more energy. However, if a certain energy has many corresponding microstates, flipping from a lower energy microstate to a corresponding microstate will happen much more frequently causing a shift towards that energy across many flips.

Of course, you might be asking, why this probability? However familiar-looking it is, it is a tad arbitrary. The probability is chosen to satisfy a condition known as *detailed balance*:

$$
p(s_j)P(s_j \rightarrow s_k) = p(s_k)(s_k \rightarrow s_j)
$$

for any microstates $s_j, s_k$. if the probability satisfies detailed balance, it is gauranteed that the samples will be sampled with respect to $p(s)$, the probabilty distribution function of the microstates:

$$p(s) = \frac{e^{-\beta H(s)}}{Z}$$

after enough iterations. A more complete exploration as to why this is the case can be found in [3]. So now iterating over each spin over the number of interations:

```python
# iterate over each temperature
for l in range(len(temperatures)):
  # for each temperature, reset the microstate to an inital, arbitrary microstate
  s = np.copy(s_0)
  # iterate for as many times we want to perform the iteration
  for iteration in range(num_iterations):
    # iterate over each spin in the lattice
    for i in range(N):
      for j in range(N):
        # compute energy difference is the spin is flipped the only difference that occurs is in the neighbours so
        delta_h = 2 * (s[i - 1, j] + s[(i + 1) % N, j] + s[i, j - 1] + s[i, (j + 1) % N]) * s[i, j]

        # check if energy difference is positive or negative
        if delta_h < 0:
          # if negative flip indiscriminately
          s[i, j] = -s[i, j]
        if delta_h >= 0:
          # if positive flip with some probability
          if np.random.rand() < np.exp(-1 * (1 / temperatures[l]) * delta_h):
            s[i, j] = -s[i, j]

    # for certain temperatures, keep track of it's evolution
    if l in temperatures_interest:
      state_evolution[temperatures_interest.index(l), iteration] = s

    # add the magnetisation after the kth iteration to the magnetisation buffer
    magnetisations[l] = magnetisations[l] + (np.sum(s) / N ** 2)

# determine average magnetisation over all iterations
magnetisations = np.abs(magnetisations / num_iterations)
```

That does it in principle, now all we need to do is plot the results!

```python
# plotting simulation results
plt.scatter(temperatures, magnetisations, label="Simulation Results")

# plotting exact Onsager solution
temperaturesExact = np.linspace(1.27, 2.269185314213022, 72)
plt.plot(temperaturesExact, (1 - (np.sinh(2 / temperaturesExact)) ** (-4)) ** (1 / 8), color="red", label="Exact Onsager Solution")
plt.plot(np.linspace(2.26918531421, 3.27, 72), np.zeros(72), color="red")

plt.title("Reduced magnetisation ($m$) against temperature ($T$)")
plt.ylabel("$m$")
plt.xlabel("$T$ in units of $J/k_b$")

plt.legend()

plt.savefig("magnetisationVsTemperatureMCMC.png")

plt.show()

```

And we can see that, at around the exact critical temperature, $2.27 J/k_b$, the reduced magnetisation suddenly spikes, indicating the phase transition once again, but this time at the correct critical temperature! For a better understanding, let us look at the final microstate at 2 3 temperatures: below $T_c$, around $T_c$ and above $T_c$.
